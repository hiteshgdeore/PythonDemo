```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
np.random.seed(0)

# Population 1
mu1 = 0
sigma1 = 1
X1 = np.random.normal(mu1, sigma1, 100)

# Population 2
mu2 = 3
sigma2 = 1
X2 = np.random.normal(mu2, sigma2, 100)
```

```python
P1 = 0.5   # Prior for class 1
P2 = 0.5   # Prior for class 2
```

```python
def bayes_classifier(x):
    p1 = norm.pdf(x, mu1, sigma1) * P1
    p2 = norm.pdf(x, mu2, sigma2) * P2
    return 1 if p1 > p2 else 2
```

```python
y1_pred = [bayes_classifier(x) for x in X1]
y2_pred = [bayes_classifier(x) for x in X2]
```

```python
x_vals = np.linspace(-4, 7, 500)

p1_vals = norm.pdf(x_vals, mu1, sigma1) * P1
p2_vals = norm.pdf(x_vals, mu2, sigma2) * P2

plt.plot(x_vals, p1_vals, label="Class 1")
plt.plot(x_vals, p2_vals, label="Class 2")
plt.axvline(x_vals[np.argmin(abs(p1_vals - p2_vals))],
            color='red', linestyle='--', label="Decision Boundary")

plt.legend()
plt.title("Bayes Classifier")
plt.show()
```

```python
# Estimate means
mean1 = np.mean(X1)
mean2 = np.mean(X2)

# Pooled variance
var1 = np.var(X1)
var2 = np.var(X2)
pooled_var = (var1 + var2) / 2

def lda_classifier(x):
    score1 = (x * mean1) / pooled_var - (mean1**2) / (2 * pooled_var)
    score2 = (x * mean2) / pooled_var - (mean2**2) / (2 * pooled_var)
    return 1 if score1 > score2 else 2
```

```python
def qda_classifier(x):
    score1 = -0.5*np.log(var1) - ((x - mean1)**2)/(2*var1)
    score2 = -0.5*np.log(var2) - ((x - mean2)**2)/(2*var2)
    return 1 if score1 > score2 else 2
```

```python
def confusion_matrix(true_class, predicted_class):
    TP = sum((true_class == 1) & (predicted_class == 1))
    TN = sum((true_class == 2) & (predicted_class == 2))
    FP = sum((true_class == 2) & (predicted_class == 1))
    FN = sum((true_class == 1) & (predicted_class == 2))
    return TP, FP, FN, TN
```

```python
true_labels = np.array([1]*100 + [2]*100)
predicted_labels = np.array(
    [bayes_classifier(x) for x in np.concatenate((X1, X2))]
)

error_rate = np.mean(true_labels != predicted_labels)
print("Misclassification Error:", error_rate)
```
