```python
#Q1)
import numpy as np

# Parameters
mu = 0
sigma = 1
n1 = 10
n2 = 50
B = 10000

def simulate_estimators(n):
    mean_est = []
    var_n = []
    var_n1 = []

    for _ in range(B):
        sample = np.random.normal(mu, sigma, n)
        mean_est.append(np.mean(sample))
        var_n.append(np.var(sample, ddof=0))   # divisor n
        var_n1.append(np.var(sample, ddof=1))  # divisor n-1

    return np.array(mean_est), np.array(var_n), np.array(var_n1)

def summarize(estimates, true_value):
    bias = np.mean(estimates) - true_value
    variance = np.var(estimates)
    mse = bias**2 + variance
    return bias, variance, mse

for n in [n1, n2]:
    mean_est, var_n, var_n1 = simulate_estimators(n)

    print(f"\nSample size n = {n}")

    print("Sample Mean:")
    print(summarize(mean_est, mu))

    print("Variance (divisor n):")
    print(summarize(var_n, sigma**2))

    print("Variance (divisor n-1):")
    print(summarize(var_n1, sigma**2))
```

```python
#(a) Bias, Variance, and MSE of Estimators
# Parameters
mu = 0
sigma = 1
sample_sizes = [10, 50]

results = {}

for n in sample_sizes:
    means = []
    var_n = []
    var_n1 = []
    
    for _ in range(B):
        x = np.random.normal(mu, sigma, n)
        means.append(np.mean(x))
        var_n.append(np.var(x, ddof=0))
        var_n1.append(np.var(x, ddof=1))
    
    means = np.array(means)
    var_n = np.array(var_n)
    var_n1 = np.array(var_n1)
    
    results[n] = {
        "mean_bias": np.mean(means) - mu,
        "mean_var": np.var(means),
        "mean_mse": np.mean((means - mu)**2),
        
        "var_n_bias": np.mean(var_n) - sigma**2,
        "var_n_var": np.var(var_n),
        "var_n_mse": np.mean((var_n - sigma**2)**2),
        
        "var_n1_bias": np.mean(var_n1) - sigma**2,
        "var_n1_var": np.var(var_n1),
        "var_n1_mse": np.mean((var_n1 - sigma**2)**2),
    }

for n, res in results.items():
    print(f"\nSample size n = {n}")
    for k, v in res.items():
        print(f"{k}: {v:.5f}")
```
