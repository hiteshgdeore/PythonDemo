```python
import numpy as np
import random
from itertools import chain

np.random.seed(42)
random.seed(42)

# =========================================================
# (a) SIMPLE RANDOM SAMPLING
# =========================================================

# Population
Y = np.array([
    12, 15, 18, 20, 22, 25, 27, 30, 32, 35,
    14, 16, 19, 21, 23, 26, 28, 31, 33, 36,
    13, 17, 24, 29, 34, 37, 38, 39, 40, 42
])

N = len(Y)
n = 8

# True population parameters
pop_mean = np.mean(Y)
pop_var = np.var(Y, ddof=0)

print("=== SIMPLE RANDOM SAMPLING ===")
print("Population Mean:", pop_mean)
print("Population Variance:", pop_var)

# --- With replacement ---
sample_wr = np.random.choice(Y, size=n, replace=True)
mean_wr = np.mean(sample_wr)
var_wr = np.var(sample_wr, ddof=1)

# --- Without replacement ---
sample_wor = np.random.choice(Y, size=n, replace=False)
mean_wor = np.mean(sample_wor)
var_wor = np.var(sample_wor, ddof=1)

print("\nSample with replacement:", sample_wr)
print("Mean:", mean_wr, "Variance:", var_wr)

print("\nSample without replacement:", sample_wor)
print("Mean:", mean_wor, "Variance:", var_wor)

# Repeated sampling
R = 1000
means_wr = []
means_wor = []

for _ in range(R):
    means_wr.append(np.mean(np.random.choice(Y, n, replace=True)))
    means_wor.append(np.mean(np.random.choice(Y, n, replace=False)))

print("\nBias (WR):", np.mean(means_wr) - pop_mean)
print("Bias (WOR):", np.mean(means_wor) - pop_mean)

print("Variance of estimator (WR):", np.var(means_wr))
print("Variance of estimator (WOR):", np.var(means_wor))


# =========================================================
# (b) STRATIFIED RANDOM SAMPLING
# =========================================================

print("\n=== STRATIFIED RANDOM SAMPLING ===")

stratum1 = np.array([10, 12, 14, 15, 16, 18, 20, 22])
stratum2 = np.array([25, 27, 28, 30, 32, 34, 35, 36])
stratum3 = np.array([40, 42, 45, 47, 48, 50, 52, 55])

strata = [stratum1, stratum2, stratum3]
Nh = np.array([len(s) for s in strata])
N = sum(Nh)
Wh = Nh / N

population = np.concatenate(strata)
true_mean = np.mean(population)

print("True Population Mean:", true_mean)

n = 6

# --- Proportional allocation ---
nh_prop = (Wh * n).astype(int)
nh_prop[-1] += n - sum(nh_prop)

# --- Equal allocation ---
nh_equal = np.array([n // 3] * 3)

def stratified_mean(strata, nh):
    sample_means = []
    for s, k in zip(strata, nh):
        sample = np.random.choice(s, k, replace=False)
        sample_means.append(np.mean(sample))
    return np.sum(Wh * sample_means)

mean_prop = stratified_mean(strata, nh_prop)
mean_equal = stratified_mean(strata, nh_equal)

print("\nProportional allocation sample sizes:", nh_prop)
print("Estimated Mean (Proportional):", mean_prop)

print("\nEqual allocation sample sizes:", nh_equal)
print("Estimated Mean (Equal):", mean_equal)


# Variance of stratified mean
def stratified_variance(strata, nh):
    var = 0
    for s, k, w in zip(strata, nh, Wh):
        Sh2 = np.var(s, ddof=1)
        var += (w**2) * (1 - k/len(s)) * (Sh2 / k)
    return var

print("\nVariance (Proportional):", stratified_variance(strata, nh_prop))
print("Variance (Equal):", stratified_variance(strata, nh_equal))


# =========================================================
# (c) CLUSTER RANDOM SAMPLING
# =========================================================

print("\n=== CLUSTER RANDOM SAMPLING ===")

clusters = {
    "C1": [10, 11, 12, 13, 14],
    "C2": [15, 16, 17, 18, 19],
    "C3": [20, 21, 22, 23, 24],
    "C4": [30, 31, 32, 33, 34],
    "C5": [35, 36, 37, 38, 39],
    "C6": [40, 41, 42, 43, 44]
}

cluster_list = list(clusters.values())
M = len(cluster_list)
cluster_sizes = np.array([len(c) for c in cluster_list])

population = np.array(list(chain.from_iterable(cluster_list)))
true_mean = np.mean(population)

print("True Population Mean:", true_mean)

# Sample m = 2 clusters
m = 2
selected_clusters = random.sample(cluster_list, m)
sample = list(chain.from_iterable(selected_clusters))

cluster_mean_est = np.mean(sample)

print("\nSelected clusters:", selected_clusters)
print("Cluster Sampling Mean Estimate:", cluster_mean_est)

# Variance of cluster estimator
cluster_means = np.array([np.mean(c) for c in cluster_list])
var_cluster = (1 - m/M) * np.var(cluster_means, ddof=1) / m

print("Variance of Cluster Mean Estimator:", var_cluster)

# Comparison with SRS (same expected sample size)
n_srs = m * cluster_sizes[0]
var_srs = np.var(population, ddof=1) * (1 - n_srs/len(population)) / n_srs

print("\nSRS Variance (same expected n):", var_srs)
```
